```python
#Grafico da seção de choque diferencial - teorico
#Autor: Camila Borges Sena
#Email para contato: camila19sena@gmail.com
import numpy as np
import matplotlib.pyplot as plt

# Definição dos parâmetros
alpha = 1/137  # Constante de estrutura fina
m_mu = 0.105  # Massa do múon (GeV/c^2)
E_cm =  35   #Energia do centro de massa
s = E_cm**2 #Energia do centro de massa em GeV
factor = (alpha**2 / 4 ) # Limite superior teórico
cos_theta = np.linspace(-0.8, 0.8, 100)  # Valores de cos(theta) para o teórico
#cos_theta_exp = np.linspace(-0.8, 0.8, 20)  # Valores discretos para dados experimentais

print(factor)

# Modelo teórico para a seção de choque diferencial
sdifferential_cross_section = (factor * (1 + cos_theta**2))*(0.389e6) #O fator de conversão para nb é

print(f"dif: {sdifferential_cross_section}")

# Plotando o gráfico
plt.figure(figsize=(8, 6))

# Curva teórica
plt.plot(cos_theta, sdifferential_cross_section, label="Previsão QED", linestyle="-", color="black")


# Configurações do gráfico
plt.xlabel(r"$\cos\theta$", fontsize=19)
plt.ylabel(r"$\frac{sd\sigma}{d\Omega}$ (GeV$^2$ $\cdot$ nb $\cdot$ ster$^{-1}$)", fontsize=19)
plt.xticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo x
plt.yticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo y
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
#Seção de choque total
#Definição dos parâmetros
alpha = 1/137  # Constante de estrutura fina
m_mu = 0.105  # Massa do múon (GeV/c^2)
E_cm =  35   # Energia do centro de massa
s_values = np.linspace(150, 2500, 3000)  # Valores de s variando de 150 a 2000 GeV^2


# Calculando a seção de choque para cada valor de s
sigma_total_values = ((4 * np.pi * alpha**2) / (3 * s_values)) * (0.389e9)

print(f"sig: {sigma_total_values}")

# Plotando o gráfico
plt.figure(figsize=(8, 6))
plt.plot(s_values, sigma_total_values, label="Seção de Choque Total ($σ_{total}$)", color="Black")

# Configurações do gráfico
plt.xlabel(r"$s$ ($\mathrm{GeV}^2$)", fontsize=19)
plt.ylabel(r"$\sigma_{total}$ (pb)", fontsize=19)
plt.xticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo x
plt.yticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo y
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# Parâmetros físicos
beta = 1/137   # Constante de estrutura fina
m_mu = 0.105   # GeV/c^2
E_cm = 35      # GeV
alpha = 0.8    # Parâmetro fracionário (0 > alpha < 1)
cos_theta = np.linspace(-0.8, 0.8, 100)
s = E_cm**2 #Energia do centro de massa em GeV

# Fator comum
factor = (beta**2 / (4 * E_cm**2))
relativistic_term = np.sqrt(1 - (m_mu**2 / E_cm**2))

print(factor)

# Modelo teórico para a seção de choque diferencial
sdifferential_cross_section = (factor * (1 + cos_theta**2))*(0.389e6) #O fator de conversão para nb é

print(f"dif: {sdifferential_cross_section}")

# Seção de choque diferencial fracionária (com dependência de alpha)
term_energy = (E_cm**(4 * (1 - alpha**2))) * (1 + cos_theta**2)
term_mass = (E_cm**(2 * (1 - 2 * alpha**2))) * (m_mu**2 * cos_theta**2 - m_mu**2)
sdifferential_fractional = factor * relativistic_term * (term_energy - term_mass) * (0.389e6)

print(f"dif2: {sdifferential_fractional}")

# Normaliza a seção de choque diferencial clássica dividindo cada valor pelo máximo,
# para que o maior ponto seja 1 e os demais fiquem proporcionais.
# Isso facilita comparar o formato da curva com outras versões (como a fracionária),
# sem que a diferença de escala esconda o comportamento angular.

#razão





sdifferential_crooss_section_norm = sdifferential_cross_section / np.max(sdifferential_cross_section)

sd= (sdifferential_crooss_section_norm / sdifferential_fractional)
print(f"sd: {sd}")

# Plotagem comparativa
plt.figure(figsize=(8, 6))
plt.plot(cos_theta, sdifferential_crooss_section_norm, label="Previsão QED", color="black", linewidth=2.2)
plt.plot(cos_theta, sdifferential_fractional, label=f"Versão Fracionária (α={alpha})", color="red", linestyle="--", linewidth=2)
#plt.plot(cos_theta, sd,label=f"razão", color= "blue", )

# Configurações do gráfico
plt.xlabel(r"$\cos\theta$", fontsize=19)
plt.ylabel(r"$\frac{sd\sigma}{d\Omega}$ (GeV$^2$ $\cdot$ nb $\cdot$ ster$^{-1}$)", fontsize=19)
plt.xticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo x
plt.yticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo y
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend
```
