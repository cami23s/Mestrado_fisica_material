```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros
Gamma = 1.0   # taxa de decaimento
omega0 = 2.0  # frequência do oscilador
t = np.linspace(0, 5, 500)  # intervalo de tempo

# Condições iniciais
a0 = 1.0      # valor inicial do operador de aniquilação
n0 = 5.0      # ocupação inicial

# Evoluções
a_t = a0 * np.exp(-Gamma * t / 2) * np.exp(-1j * omega0 * t)  # operador de aniquilação
n_t = n0 * np.exp(-Gamma * t)                                 # número médio de ocupação

# Figura
plt.figure(figsize=(10,5))

# Módulo de <a(t)>
plt.subplot(1,2,1)
plt.plot(t, np.abs(a_t), label=r'$|\langle a(t) \rangle|$')
plt.xlabel('Tempo')
plt.ylabel('Amplitude')
plt.title(r'Decaimento de $\langle a(t) \rangle$')
plt.grid(True)
plt.legend()

# Número médio de ocupação
plt.subplot(1,2,2)
plt.plot(t, n_t, color='orange', label=r'$n(t)$')
plt.xlabel('Tempo')
plt.ylabel('Número médio de ocupação')
plt.title(r'Decaimento de $n(t)$')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros
Gamma = 1.0   # taxa de decaimento
omega0 = 2.0  # frequência do oscilador
t = np.linspace(0, 5, 500)  # intervalo de tempo

# Condições iniciais
a0 = 1.0      # valor inicial do operador de aniquilação
n0 = 5.0      # ocupação inicial

# Evoluções
a_t = a0 * np.exp(-Gamma * t / 2) * np.exp(-1j * omega0 * t)  # operador de aniquilação
n_t = n0 * np.exp(-Gamma * t)                                 # número médio de ocupação

# Pontos importantes
t_points = [0, 2.5, 5]
a_points = np.abs(a0 * np.exp(-Gamma * np.array(t_points) / 2))
n_points = n0 * np.exp(-Gamma * np.array(t_points))

# Figura
plt.figure(figsize=(8,5))

# Curvas sobrepostas
plt.plot(t, np.abs(a_t), label=r'$|\langle a(t) \rangle|$', color='blue', linewidth=2)
plt.plot(t, n_t, label=r'$n(t)$', color='orange', linewidth=2)

# Marcas nos pontos importantes
plt.scatter(t_points, a_points, color='blue', marker='o')
plt.scatter(t_points, n_points, color='orange', marker='s')

# Anotações
for i, tp in enumerate(t_points):
    plt.text(tp, a_points[i]+0.1, f'{a_points[i]:.2f}', color='blue', ha='center')
    plt.text(tp, n_points[i]+0.3, f'{n_points[i]:.2f}', color='orange', ha='center')

# Configurações do gráfico
plt.xlabel('Tempo')
plt.ylabel('Amplitude / Ocupação')
plt.title(r'Decaimento de $|\langle a(t) \rangle|$ e $n(t)$')  # string bruta r"..."
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
