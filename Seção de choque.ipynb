```python
#Grafico da seção de choque diferencial - teorico
#Autor: Camila Borges Sena
#Email para contato: camila19sena@gmail.com
import numpy as np
import matplotlib.pyplot as plt

# Definição dos parâmetros
alpha = 1/137  # Constante de estrutura fina
m_mu = 0.105  # Massa do múon (GeV/c^2)
E_cm =  35   #Energia do centro de massa
s = E_cm**2 #Energia do centro de massa em GeV
factor = (alpha**2 / 4 ) # Limite superior teórico
cos_theta = np.linspace(-0.8, 0.8, 100)  # Valores de cos(theta) para o teórico
#cos_theta_exp = np.linspace(-0.8, 0.8, 20)  # Valores discretos para dados experimentais

print(factor)

# Modelo teórico para a seção de choque diferencial
sdifferential_cross_section = (factor * (1 + cos_theta**2))*(0.389e6) #O fator de conversão para nb é

print(f"dif: {sdifferential_cross_section}")

# Plotando o gráfico
plt.figure(figsize=(8, 6))

# Curva teórica
plt.plot(cos_theta, sdifferential_cross_section, label="Previsão QED", linestyle="-", color="black")


# Configurações do gráfico
plt.xlabel(r"$\cos\theta$", fontsize=19)
plt.ylabel(r"$\frac{sd\sigma}{d\Omega}$ (GeV$^2$ $\cdot$ nb $\cdot$ ster$^{-1}$)", fontsize=19)
plt.xticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo x
plt.yticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo y
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
#Seção de choque total
#Definição dos parâmetros
alpha = 1/137  # Constante de estrutura fina
m_mu = 0.105  # Massa do múon (GeV/c^2)
E_cm =  35   # Energia do centro de massa
s_values = np.linspace(150, 2500, 3000)  # Valores de s variando de 150 a 2000 GeV^2


# Calculando a seção de choque para cada valor de s
sigma_total_values = ((4 * np.pi * alpha**2) / (3 * s_values)) * (0.389e9)

print(f"sig: {sigma_total_values}")

# Plotando o gráfico
plt.figure(figsize=(8, 6))
plt.plot(s_values, sigma_total_values, label="Seção de Choque Total ($σ_{total}$)", color="Black")

# Configurações do gráfico
plt.xlabel(r"$s$ ($\mathrm{GeV}^2$)", fontsize=19)
plt.ylabel(r"$\sigma_{total}$ (pb)", fontsize=19)
plt.xticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo x
plt.yticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo y
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# Parâmetros físicos
beta = 1/137   # Constante de estrutura fina
m_mu = 0.105   # GeV/c^2
E_cm = 35      # GeV
alpha = 0.8   # Parâmetro fracionário (0 > alpha < 1)
alpha1 = 0.6 # Parâmetro fracionário (0 > alpha < 1)
alpha2 = 0.95
#alpha2 = 1.00
cos_theta = np.linspace(-0.8, 0.8, 100)
s = E_cm**2 #Energia do centro de massa em GeV

# Fator comum
factor = (beta**2 / (4 * E_cm**2))
relativistic_term = np.sqrt(1 - (m_mu**2 / E_cm**2))

print(factor)

# Modelo teórico para a seção de choque diferencial
sdifferential_cross_section = (factor * (1 + cos_theta**2))*(0.389e6) #O fator de conversão para nb é

print(f"dif: {sdifferential_cross_section}")

# Seção de choque diferencial fracionária (com dependência de alpha)
term_energy = (E_cm**(4 * (1 - alpha**2))) * (1 + cos_theta**2)
term_mass = (E_cm**(2 * (1 - 2 * alpha**2))) * (m_mu**2 * cos_theta**2 - m_mu**2)
sdifferential_fractional = factor * relativistic_term * (term_energy - term_mass) * (0.389e6)

# Seção de choque diferencial fracionária (com dependência de alpha1)
term_energy = (E_cm**(4 * (1 - alpha1**2))) * (1 + cos_theta**2)
term_mass = (E_cm**(2 * (1 - 2 * alpha1**2))) * (m_mu**2 * cos_theta**2 - m_mu**2)
sdifferential_fractional1 = factor * relativistic_term * (term_energy - term_mass) * (0.389e6)

# Seção de choque diferencial fracionária (com dependência de alpha2)
term_energy = (E_cm**(4 * (1 - alpha2**2))) * (1 + cos_theta**2)
term_mass = (E_cm**(2 * (1 - 2 * alpha2**2))) * (m_mu**2 * cos_theta**2 - m_mu**2)
sdifferential_fractional2 = factor * relativistic_term * (term_energy - term_mass) * (0.389e6)

print(f"dif2: {sdifferential_fractional}")


#razão

sd= (sdifferential_fractional / sdifferential_cross_section)
print(f"sd: {sd}")

media = np.mean(sd)
print("Média:", media)
desvio = np.std(sd)
print("Desvio padrão:", desvio)

sd1= (sdifferential_fractional1 / sdifferential_cross_section)
print(f"sd1: {sd1}")
media1 = np.mean(sd1)
print("Média1:", media1)
desvio1 = np.std(sd1)
print("Desvio padrão1:", desvio1)


sd2= (sdifferential_fractional2 / sdifferential_cross_section)
print(f"sd2: {sd2}")
media2 = np.mean(sd2)
print("Média2:", media2)
desvio2 = np.std(sd2)
print("Desvio padrão2:", desvio2)

# Plotagem comparativa
plt.figure(figsize=(8, 6))
plt.plot(cos_theta, sdifferential_cross_section, label="QED", color="black", linewidth=2.2)
plt.plot(cos_theta, sdifferential_fractional1, label="Versão Fracionária (α=0.6)", color="blue", linewidth=2.2)
plt.plot(cos_theta, sdifferential_fractional2, label="Versão Fracionária (α=0.95)", color="green", linewidth=2.2)
plt.plot(cos_theta, sdifferential_fractional, label=f"Versão Fracionária (α={alpha})", color="red", linestyle="--", linewidth=2)
#plt.plot(cos_theta, sd,label=f"razão", color= "pink", )

# Configurações do gráfico
plt.xlim(-1, 1)  # define os limites do eixo x
plt.xlabel(r"$\cos\theta$", fontsize=19)
plt.ylabel(r"$\frac{sd\sigma}{d\Omega}$ (GeV$^2$ $\cdot$ nb $\cdot$ ster$^{-1}$)", fontsize=19)
plt.xticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo x
plt.yticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo y
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)
#plt.legend
plt.legend(fontsize=12, loc="upper center", bbox_to_anchor=(0.5, -0.1), ncol=2)



# Escalas logarítmicas
#plt.xscale("-1, 1")   # eixo x em log
plt.yscale("log")   # eixo y em log
```

```python
# Parâmetros físicos
beta = 1/137
m_mu = 0.105
alpha = 0.8
alpha1=0.6
alpha2=0.95
s_values = np.linspace(100, 2000, 3000)
E_values = np.sqrt(s_values)  # Energia variável conforme s

# Seção de choque total clássica (QED)
sigma_total_classic = ((4 * np.pi * beta**2) / (3 * s_values)) * (0.389e9)

print(f"sig:{sigma_total_classic}")


# Seção de choque total fracionária (dependente de s)
relativistic_term = np.sqrt(1 - (m_mu**2 / E_values**2))
term_energy = E_values**(4 * (1 - alpha**2))
term_mass = 0.5 * m_mu**2 * E_values**(2 * (1 - 2 * alpha**2))
sigma_total_fractional = ((4 * np.pi * beta**2) / (3 * E_values**2)) * relativistic_term * (term_energy + term_mass) * (0.389e9)

# Seção de choque total fracionária (dependente de s)1
relativistic_term = np.sqrt(1 - (m_mu**2 / E_values**2))
term_energy = E_values**(4 * (1 - alpha1**2))
term_mass = 0.5 * m_mu**2 * E_values**(2 * (1 - 2 * alpha1**2))
sigma_total_fractional1 = ((4 * np.pi * beta**2) / (3 * E_values**2)) * relativistic_term * (term_energy + term_mass) * (0.389e9)

# Seção de choque total fracionária (dependente de s)2
relativistic_term = np.sqrt(1 - (m_mu**2 / E_values**2))
term_energy = E_values**(4 * (1 - alpha2**2))
term_mass = 0.5 * m_mu**2 * E_values**(2 * (1 - 2 * alpha2**2))
sigma_total_fractional2 = ((4 * np.pi * beta**2) / (3 * E_values**2)) * relativistic_term * (term_energy + term_mass) * (0.389e9)


# Configurações do gráfico
plt.figure(figsize=(8, 6))
plt.plot(s_values, sigma_total_classic, color="black", linewidth=2, label="QED")
plt.plot(s_values, sigma_total_fractional, color="red", linestyle="--", linewidth=2, label=f"Fracionária (α={alpha})")
plt.plot(s_values, sigma_total_fractional2, color="green", linewidth=2, label="Fracionária (α=0.95)")
plt.plot(s_values, sigma_total_fractional1, color="blue", linewidth=2, label="Fracionária (α=0.6)")


plt.xlabel(r"$sGeV^2$", fontsize=19)
plt.ylabel(r"$\sigma_{total}$", fontsize=19)
plt.xticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo x
plt.yticks(fontsize=14)  # Aumenta o tamanho da fonte dos números do eixo y
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)
#plt.legend
plt.legend(fontsize=12, loc="upper center", bbox_to_anchor=(0.5, -0.1), ncol=2)



# Escalas logarítmicas
plt.xscale("log")   # eixo x em log
plt.yscale("log")   # eixo y em log
```
